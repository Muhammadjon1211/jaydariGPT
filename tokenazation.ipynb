```python
text = "Hello world this is a simple example to show how tokenazation works in NLP and a sentance" # 17 token
print(text)
```

```python
tokens = text.split()
print(tokens)
```

```python
vocab = {}
current_id = 0

for token in tokens:
  if token not in vocab:
    vocab[token] = current_id
    current_id += 1

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]
print(encoded)
```

```python
reverse_vocab = {id: token for token, id in vocab.items()}
decoded = " ".join([reverse_vocab[id] for id in encoded])
print(decoded)
#print(reverse_vocab)
```

```python
# Embedding vector => dense numerical array | Vector DB(store embedding vectors)
# "hello" => [0.41, 0.22, -0.77, 0.56, ......]
```

```python
import torch
import torch.nn as nn

embedding_dim = 8
vocab_size = len(vocab)

embedding_layer = nn.Embedding(vocab_size, embedding_dim)

input_ids = torch.tensor([encoded])
embedding = embedding_layer(input_ids)

print(embedding)
print(embedding.shape)
```

```python
from transformers import AutoTokenizer

model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
print('Vocab size:', tokenizer.vocab_size)
print("Special tokens:", tokenizer.special_tokens_map)
```

```python
text = "Hello how are you doing?"
print("text:", text)
```

```python
tokens = tokenizer.tokenize(text)
print("tokens:", tokens)
```

```python
ids =tokenizer.convert_tokens_to_ids(tokens)
print("ids:", ids)
```

```python
encoded = tokenizer(text)
print("encoded:", encoded)
```

```python
print("Input IDs:", encoded["input_ids"])
print("Attention Mask real data:", encoded["attention_mask"])
```

```python
"Hi" #=> [1,345, 2, 2, 2] // length 5
"How are you?" #=> [1, 345, 23, 432, 2347] // length 5
# "attention_mask": [1,1,0,0,0]
```

```python
decoded_text = tokenizer.decode(encoded["input_ids"])
print("decoded_text:", decoded_text)
```

```python
texts = [
    "Hello, how are you?",
    "Today we are learning about tokenizers.",
    "TinyLlama is a small but powerfull language model."
]

batch = tokenizer(texts,
                  padding=True,
                  truncation=True,
                  return_tensors="pt"
                  )

print(batch.keys())

print("Input IDs shape:", batch["input_ids"].shape)
print("Attention Mask shape:", batch["attention_mask"].shape)
```
