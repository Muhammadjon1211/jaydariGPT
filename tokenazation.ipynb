```python
text = "Hello world this is a simple example to show how tokenazation works in NLP and a sentance" # 17 token
print(text)
```

```python
tokens = text.split()
print(tokens)
```

```python
vocab = {}
current_id = 0

for token in tokens:
  if token not in vocab:
    vocab[token] = current_id
    current_id += 1

print(vocab)
```

```python
encoded = [vocab[token] for token in tokens]
print(encoded)
```

```python
reverse_vocab = {id: token for token, id in vocab.items()}
decoded = " ".join([reverse_vocab[id] for id in encoded])
print(decoded)
#print(reverse_vocab)
```

```python
# Embedding vector => dense numerical array | Vector DB(store embedding vectors)
# "hello" => [0.41, 0.22, -0.77, 0.56, ......]
```

```python
import torch
import torch.nn as nn

embedding_dim = 8
vocab_size = len(vocab)

embedding_layer = nn.Embedding(vocab_size, embedding_dim)

input_ids = torch.tensor([encoded])
embedding = embedding_layer(input_ids)

print(embedding)
print(embedding.shape)
```
